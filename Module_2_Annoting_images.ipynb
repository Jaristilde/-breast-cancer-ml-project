```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
import requests

image_url = 'https://raw.githubusercontent.com/Jaristilde/cv2/main/Images/Apollo-8-Launch.jpg'
image_filename = 'Apollo-8-Launch.jpg'

response = requests.get(image_url)
if response.status_code == 200:
    with open(image_filename, 'wb') as f:
        f.write(response.content)
    print(f"Image '{image_filename}' downloaded successfully.")
else:
    print(f"Failed to download image. Status code: {response.status_code}")
```

```python
# Read image as color (3 channels)
image = cv2.imread('Apollo-8-Launch.jpg', cv2.IMREAD_COLOR)

# Display the original image
plt.figure(figsize=(7, 7))
plt.imshow(image)
plt.show()

# Check image shape (height, width, channels)
print(image.shape) # Should show 3 channels
```

```python
# Convert to grayscale
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Display the grayscale image
plt.figure(figsize=(7, 7))
plt.imshow(gray_image, cmap='gray')
plt.show()

# Check image shape (height, width)
print(gray_image.shape) # Should show 2 channels
```

```python
# Apply a binary threshold
# Pixels above 127 become 255 (white), below 127 become 0 (black)
ret, binary_image = cv2.threshold(gray_image, 127, 255, cv2.THRESH_BINARY)

# Display the binary image
plt.figure(figsize=(7, 7))
plt.imshow(binary_image, cmap='gray')
plt.show()

# Check image shape (height, width)
print(binary_image.shape)
```
